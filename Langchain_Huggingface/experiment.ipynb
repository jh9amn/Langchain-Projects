#### Huggingface Endpoints

##### How to access huggingface Models with API

There are also two ways to use this class. You can specify the model with the repo_id paramter. Those endpoints use the serverless API, which is particularly beneficial to people using pro accounts or enterprise hub. Still, regular usera can already have access to fair amount if request by connnecting with their HF token in the environment where they are executinh the code..

```python
from langchain_huggingface import HuggingFaceEndpoint
```

```python

# repo_id = "mistralai/Mistral-7B-Instruct-v0.3"
repo_id = "google/gemma-3-4b-it"
# repo_id = "google/flan-t5-large"

llm = HuggingFaceEndpoint(
    repo_id=repo_id,
    task="conversational",
    huggingfacehub_api_token=os.getenv("HF_TOKEN"),
    max_new_tokens=150,
    temperature=0.7
)

llm
```

```python

response = llm.invoke("What is machine learning?")
print(response)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_classic.chains.llm import LLMChain

template = """
Question: {question}
Answer: Lets think step by step.
"""

prompt = PromptTemplate(
    template = template,
    input_variable=["question"]
)

print(prompt)
```

```python
llm_chain = LLMChain(llm = llm, prompt = prompt)
llm.invoke("What's machine leanring ?")
```

#### Open Source embeddings

https://docs.langchain.com/oss/python/integrations/embeddings/bge_huggingface

```python
from langchain_community.embeddings import HuggingFaceBgeEmbeddings

model_name = "BAAI/bge-small-en"
model_kwargs = {"device": "cpu"}
encode_kwargs = {"normalize_embeddings": True}
hf = HuggingFaceBgeEmbeddings(
    model_name=model_name, model_kwargs=model_kwargs, encode_kwargs=encode_kwargs
)
```

```python
embedding = hf.embed_query("hi this is harrison")
len(embedding)
```
